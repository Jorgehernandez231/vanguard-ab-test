## Clean Process – df_final_demo <a id="clean-process--df_final_demo"></a>

[↑ Back to top](#top)


```python
# Load the dataset
df_final_demo = pd.read_csv('F:/Ironhack/vanguard-ab-test/Data/Raw/df_final_demo.txt')
```

```python
# Clean the dataset using the basic_data_cleaning function
# Here, we choose to drop rows with missing values because the missing data is minimal
clean_df_final_demo = basic_data_cleaning(df_final_demo, drop_missing=True, fill_missing=0, remove_outliers=False, drop_duplicates_subset=None )
```

```python
# Convert specific columns to integer type
clean_df_final_demo["clnt_tenure_yr"] = clean_df_final_demo["clnt_tenure_yr"].astype(int)
clean_df_final_demo["clnt_tenure_mnth"] = clean_df_final_demo["clnt_tenure_mnth"].astype(int)
clean_df_final_demo["num_accts"] = clean_df_final_demo["num_accts"].astype(int)
clean_df_final_demo["calls_6_mnth"] = clean_df_final_demo["calls_6_mnth"].astype(int)
clean_df_final_demo["logons_6_mnth"] = clean_df_final_demo["logons_6_mnth"].astype(int)
# Round the 'bal' column to 2 decimal places
clean_df_final_demo['bal'] = clean_df_final_demo['bal'].round(2)
```

```python
# Load the dataset 'df_final_web_data_pt_1.txt'
df_final_web_data_pt_1 = pd.read_csv('F:/Ironhack/vanguard-ab-test/Data/Raw/df_final_web_data_pt_1.txt')
```

```python
# Clean the dataset using the basic_data_cleaning function
# Here, we choose to fill missing values with 'None' because the missing data is significant and dropping rows would lead to substantial data loss.
# We also choose not to remove outliers or drop duplicates in this case.
# The 'drop_duplicates_subset' parameter is set to None to avoid dropping any duplicates. 

clean_df_final_web_data_pt_1 = basic_data_cleaning(df_final_web_data_pt_1, drop_missing=False, fill_missing='None', remove_outliers=False, drop_duplicates_subset=None )
```

### df_final_web_data_pt_2 <a id="df_final_web_data_pt_2"></a>

```python
# Load the dataset 'df_final_web_data_pt_2.txt'
df_final_web_data_pt_2 = pd.read_csv('F:/Ironhack/vanguard-ab-test/Data/Raw/df_final_web_data_pt_2.txt')
```

```python
# Clean the dataset using the basic_data_cleaning function
# Here, we choose to fill missing values with 'None' because the missing data is significant and dropping rows would lead to substantial data loss.
# We also choose not to remove outliers or drop duplicates in this case.
# The 'drop_duplicates_subset' parameter is set to None to avoid dropping any duplicates. 
clean_df_final_web_data_pt_2 = basic_data_cleaning(df_final_web_data_pt_2, drop_missing=False, fill_missing='None', remove_outliers=False, drop_duplicates_subset=None )
```
